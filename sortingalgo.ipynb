```python
#selection sort


n= int(input())
lst= list(map(int,input().split()))



def select(lst):
  for i in range(n):
      min= i
      for j in range(i+1, n):
         if lst[j]<lst[min]:

            min= j
      lst[i],lst[min]<lst[min],lst[i]
  return lst


select(lst)
```

```python
#buble sort

n= int(input())
lst= list(map(int,input().split()))

def buble(lst):
  for i in range(n-1):
    for j in range(n-i-1):
      if lst[j]>lst[j+1]:
        lst[j], lst[j+1] = lst[j+1], lst[j]
  return lst

print("list setelah buble sort",buble(lst))
```

```python
#insertion sort


n= int(input())
lst= list(map(int,input().split()))



for i in range(1,n):
  kunci=lst[i]
  j= i-1

  while j >= 0 and lst[j] > kunci:
    lst[j + 1] = lst[j]
    j -= 1
# sisipkan ke posisi yang benar
    lst[j + 1] = kunci


print(lst)
```
